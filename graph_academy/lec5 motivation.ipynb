### Schema
When we define a LangGraph, we use a state schema  

The state schema represents the structure and types of data that our graph will use  

All nodes are expected to communicate with that scchema.  

LangGraph offers flexibility in how you define state schema, accommodating various Python types and validation approaches.  

Like TypeDicted


```python
from typing import TypedDict, Annotated, List
from langchain_core.messages import AnyMessage
from langgraph.graph.message import add_messages
from langgraph.graph import MessagesState
## MessagesState等价写法
class CustomMessagesState(TypedDict):
    messages: Annotated[List[AnyMessage], add_messages]
    added_key_1 : str
    added_key_2 : str

## 等价写法
class ExtendedMessagesState(MessagesState):
    added_key_1 : str
    added_key_2 : str
    
    
```
